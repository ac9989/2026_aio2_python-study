# 연습문제

```python
snack_box = ['오예스']
snack_box
```

```python
x = input("먹고 싶은 간식을 입력하세요: ")
snack_box.append(x)
snack_box
```

```python
snack_box = ['오예스']
x = input("먹고 싶은 간식을 입력하세요: ").split(', ')
snack_box += x
snack_box
```

```python
# 몇 상자 주문하시겠어요?
order_count = int(input("몇 상자 주문하시겠어요? "))
```

```python
# 제외하고 싶은 간식이 있으면 번호를 입력하세요
snack_num = int(input("제외하고 싶은 간식의 번호를 입력하세요: "))
snack_box.remove(snack_box[snack_num-1])
```

```python
print(f"주문하신 간식은 {snack_box}이며, 총 {order_count}상자 주문되었습니다.")
```
