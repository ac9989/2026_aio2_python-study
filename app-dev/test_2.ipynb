```python
# 예제 URL (안정적인 사이트 사용 권장)
target_url = "https://www.python.org/"
print(f"--- URL GET 요청 시작: {target_url} ---")
try:
# with 문을 사용하여 자동으로 리소스 정리
    with urllib.request.urlopen(target_url, timeout=10) as response:
        status = response.status
        print(f"상태 코드: {status}")

        headers = response.headers
        print(f"콘텐츠 타입: {headers.get('Content-Type', '알 수 없음')}")

        # 헤더에서 인코딩 정보 얻기 (없으면 utf-8 가정)
        encoding = headers.get_content_charset() or 'utf-8'
        print(f"감지된/가정된 인코딩: {encoding}")
        if 200 <= status < 300: # 성공적인 응답 확인
            html_bytes = response.read() # 본문을 바이트로 읽기
            # 바이트를 문자열로 디코딩
            html_content = html_bytes.decode(encoding)
            print("\n--- 응답 내용 (일부) ---")
            print(html_content[:500] + "...") # 앞 500자만 출력
        else:print(f"오류 응답 수신: 상태 코드 {status}")

except urllib.error.HTTPError as e:
    # HTTP 오류 처리 (예: 404, 500)
    print(f"HTTP 오류 발생: {e.code} {e.reason}")
except urllib.error.URLError as e:
# URL 관련 오류 처리 (예: 호스트 찾을 수 없음, 네트워크 연결 불가)
    print(f"URL 오류 발생: {e.reason}")
except TimeoutError:
    print("요청 시간 초과")
except Exception as e:
    print(f"기타 오류 발생: {e}")
```

```python
import requests
import json # JSONDecodeError 처리용
import datetime # 시간 정보 활용

print(f"--- requests 라이브러리 예제 (시각: {datetime.datetime.now().strftime('%H:%M:%S KST')}) ---")

# ==========================================
# 1. 간단한 GET 요청 (예외 처리 포함) [3, 4]
# ==========================================
print("\n1. 간단한 GET 요청 (httpbin.org 사용)")
try:
    response = requests.get('https://httpbin.org/get', timeout=5)
    response.raise_for_status() # 응답 상태 코드가 4xx 또는 5xx 오류일 경우 예외 발생 [11]
    print(f"  상태 코드: {response.status_code}")  
    print(f"  응답 OK?: {response.ok}")  
    print(f"  콘텐츠 타입: {response.headers.get('content-type')}")  
    
    # JSON 응답 처리
    try:
        data = response.json() # 응답 본문을 파이썬 딕셔너리로 자동 변환 [4, 11]
        print(f"  JSON 응답에서 'origin' IP: {data.get('origin')}") 
    except requests.exceptions.JSONDecodeError:
        print("  응답이 JSON 형식이 아닙니다.") 
        
except requests.exceptions.Timeout:
    print("  요청 시간 초과") 
except requests.exceptions.HTTPError as errh:
    print(f"  HTTP 오류 발생: {errh}") 
except requests.exceptions.RequestException as err:
    print(f"  요청 관련 오류 발생: {err}") 

# ==========================================
# 2. GET 요청 (쿼리 파라미터 포함) [4, 5]
# ==========================================
print("\n2. GET 요청 (쿼리 파라미터 포함)")
# 딕셔너리를 전달하면 requests가 알아서 URL 인코딩을 처리해 줍니다 [4, 12]
payload = {'search_term': '서울 날씨', 'region': 'kr'} 
try:
    response_search = requests.get('https://httpbin.org/get', params=payload, timeout=5) 
    # print(f"  서버가 받은 파라미터: {response_search.json().get('args')}")
except requests.exceptions.RequestException as e:
    print(f"  파라미터 포함 GET 오류: {e}") 

# ==========================================
# 3. POST 요청 (폼 데이터 전송) [5, 6]
# ==========================================
print("\n3. POST 요청 (폼 데이터)")
form_payload = {'user_id': 'seoul_citizen', 'action': 'update_profile'} 
try:
    # data 인자에 딕셔너리를 전달하면 form-encoded 데이터로 전송됩니다 [2, 5]
    response_post_form = requests.post('https://httpbin.org/post', data=form_payload, timeout=5) 
    response_post_form.raise_for_status() 
    print(f"  POST(폼) 상태 코드: {response_post_form.status_code}") 
except requests.exceptions.RequestException as e:
    print(f"  POST(폼) 오류: {e}") 

# ==========================================
# 4. POST 요청 (JSON 데이터 전송) [6, 7]
# ==========================================
print("\n4. POST 요청 (JSON 데이터)")
json_payload = {
    'log_level': 'INFO',
    'message': 'User login successful',
    'timestamp': datetime.datetime.now().isoformat(),
    'source_ip': '127.0.0.1'
}
try:
    # json 인자를 사용하면 딕셔너리가 JSON으로 자동 변환되고 Content-Type: application/json 헤더가 설정됩니다 [6, 7, 13]
    response_post_json = requests.post('https://httpbin.org/post', json=json_payload, timeout=5)
    response_post_json.raise_for_status() 
    print(f"  POST(JSON) 상태 코드: {response_post_json.status_code}") 
except requests.exceptions.RequestException as e:
    print(f"  POST(JSON) 오류: {e}") 

# ==========================================
# 5. 사용자 정의 헤더 추가 [7, 8]
# ==========================================
print("\n5. 사용자 정의 헤더 추가")
custom_headers = {'User-Agent': 'MySeoulApp/1.0', 'Accept-Language': 'ko-KR'} 
try:
    # headers 인자에 딕셔너리로 정의한 커스텀 헤더를 넘겨줍니다 [2, 7]
    response_headers = requests.get('https://httpbin.org/headers', headers=custom_headers, timeout=5)
    print("  사용자 정의 헤더를 포함한 요청 성공.") 
except requests.exceptions.RequestException as e:
    print(f"  헤더 포함 요청 오류: {e}") 

# ==========================================
# 6. 세션(Session) 사용 예제 [8-10]
# ==========================================
print("\n--- requests.Session 사용 예제 ---")
# 세션 객체는 쿠키를 자동으로 유지하고 TCP 연결을 재사용(Connection Pooling)합니다 [8, 9]
with requests.Session() as s:
    # 세션 기본 헤더 설정 (이 세션을 통해 보내는 모든 요청에 자동으로 적용됨) [8, 9]
    s.headers.update({'X-Client-Info': 'SeoulClient/v2'}) 
    try:
        # 세션을 통해 GET 요청 
        print("  세션 GET 요청...")
        resp1 = s.get('https://httpbin.org/get', timeout=5) 
        resp1.raise_for_status() 
        print("    세션 GET 성공.") 
        
        # 세션을 통해 POST 요청 
        print("  세션 POST 요청...")
        resp2 = s.post('https://httpbin.org/post', data={'session_data': 'true'}, timeout=5)
        resp2.raise_for_status()
        print("    세션 POST 성공.")
    except requests.exceptions.RequestException as e:
        print(f"  세션 사용 중 오류: {e}")
```
