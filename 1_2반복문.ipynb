# 반복문 개요

파이썬의 반복문은 크게 for 문과 while 문 두 가지로 나뉜다. 
for 문은 반복 횟수가 정해진 경우에 사용한다.
while 문은 조건이 참인 동안 계속 반복할 때 사용한다.

## for 문

```python
fruits = ["사과", "바나나", "포도"]

for fruit in fruits:  #시퀀스
    print(fruit)
```

```python
hi = "hello"
for i in hi:  #시퀀스
    print(i)
```

```python
for x in range(5):  #시퀀스
    print(x)
```

```python
user_input = ""

while user_input != "exit":
    user_input = input("원하는 과자를 입력하세요 (종료하려면 'exit' 입력): ")
    print(user_input)
```
