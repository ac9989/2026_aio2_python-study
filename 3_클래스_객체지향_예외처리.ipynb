## class calcurator 정의

### 문제1
cal1 = Calculator([1,2,3,4,5])  
cal1.sum() # 15  
cal1.avg() # 3.0

```python
class Calculator:
    def __init__(self, num_list):
        self.num_list = num_list
    def sum(self):
        return sum(self.num_list)
    def avg(self):
        return sum(self.num_list) / len(self.num_list)
```

```python
cal1 = Calculator([1,2,3,4,5])  
print(cal1.sum()) # 15  
print(cal1.avg()) # 3.0

cal2 = Calculator([6,7,8,9,10])  
print(cal2.sum()) # 40  
print(cal2.avg()) # 8.0
```

은행이름은 SC은행으로 계좌번호는 3자리-2자리-6자리 형태로 랜덤하게 생성됩니다.

```python
import random

class Account:
    def __init__(self, owner, balance):
        self.bank = "SC은행"
        self.owner = owner
        self.balance = balance

        first = random.randint(0, 999)
        second = random.randint(0, 99)
        third = random.randint(0, 999999)

        self.account_number = f"{first:03d}-{second:02d}-{third:06d}"
```

```python
import random

class Account:
    bank_name = 'sc은행'  # 클래스 변수

    def __init__(self, name, balance):
        self.name = name  # 인스턴스 변수
        self.balance = balance
        self.account_no = Account.make_account_no()

    @staticmethod
    def make_account_no():
        num1 = random.randint(0,999)
        num2 = random.randint(0,99)
        num3 = random.randint(0,999999)

        return f"{num1:03d}-{num2:02d}-{num3:06d}"


    def print_info(self,):
        return f'은행이름 : {Account.bank_name} \n 계좌번호 : {self.account_no}'

    def 입금(self, amount):
        self.balance += amount

    def 출금(self, amount):
        self.balance -= amount

    def 잔고조회(self):
        return f'{self.name}님의 잔고는 {self.balance} 입니다.'
```

```python
ac1 = Account('홍길동',2000)
print(ac1.print_info())
print(ac1.잔고조회())
print(ac1.make_account_no())
```

# 예외와 오류, 예외처리

오류(error)는 문법적으로 틀린 것. 실행이 되지 않으며, 반드시 수정해야만 실행이 된다.  
예외는 문법적으로 틀리지는 않았지만, 실 사용에는 큰 불편을 주는 문제이다. (0으로 나누기, 존재하지 않는 파일 열기 등)

```python
try:
    num = int(input("숫자를 입력하세요: "))
    result = 10 / num
    print(result)
except ZeroDivisionError:           #0으로 나누기 에러
    print("0으로 나눌 수 없습니다.")
except ValueError:                  # 문자열을 정수로 변환 시 에러 처리
    print("정수를 입력해주세요.")
```

## 예외값 처리

```python
class AgeException(Exception):
    def __init__(self, msg):
        self.message = msg
```

```python
def input_age():
    #나이를 입력
    age = int(input('나이를 입력하세요: '))
    

    #입력값 체크
    if age < 0:  #나이는 양수
        raise AgeException('나이는 양수로 입력해주세요!')
    elif age > 150:  #상한 체크
        raise AgeException('나이를 잘못 입력하신 것 같아요!')
    else :
        return age

# 예외 처리를 하는 나이 입력 구현
try:
    age = input_age()
except AgeException as ae:
    print(ae.message)
except ValueError:                
    print("자연수로 입력해주세요!")
else :
    print(f"당신의 나이는 {age}살 이군요!")
```
