## 1. math — 수학 연산
python  
import math  

math.sqrt(16)        # 4.0 (제곱근)  
math.pow(2, 3)        # 8.0 (거듭제곱, ** 연산자와 유사)  
math.floor(3.7)       # 3 (내림)  
math.ceil(3.2)         # 4 (올림)  
math.trunc(3.9)        # 3 (소수점 버림, floor와 음수에서 다름)  
math.pi                 # 3.141592... (원주율 상수)  
math.e                   # 2.718281... (자연상수)  
math.log(100, 10)     # 2.0 (로그, 밑 생략시 자연로그)  
math.factorial(5)     # 120 (5!)  
math.gcd(12, 18)      # 6 (최대공약수)  
math.hypot(3, 4)      # 5.0 (빗변 길이, 유클리드 거리)  

## 2. random — 난수 생성

## 3. datetime — 날짜/시간 처리  
python  
from datetime import datetime, date, timedelta  

datetime.now()                     # 현재 날짜+시간  
date.today()                        # 오늘 날짜만  
datetime(2025, 1, 1)             # 특정 날짜 객체 생성  

now = datetime.now()  
now.strftime("%Y-%m-%d %H:%M")   # 문자열로 포맷팅 → "2026-08-10 14:30"  
datetime.strptime("2025-01-01", "%Y-%m-%d")  # 문자열 → datetime 객체  
  
today = date.today()  
today + timedelta(days=7)         # 7일 후 날짜  
today - timedelta(days=1)          # 어제 날짜  

## 4. os — 운영체제/파일시스템  
python  
import os  

os.getcwd()                    # 현재 작업 디렉토리 경로  
os.listdir(".")                # 폴더 내 파일/폴더 목록  
os.path.exists("file.txt")     # 파일/폴더 존재 여부  
os.path.join("a", "b.txt")     # 경로 합치기 (OS별 구분자 자동처리) → "a/b.txt"  
os.path.abspath("file.txt")    # 절대경로로 변환  
os.path.basename("/a/b.txt")   # "b.txt" (파일명만 추출)  
os.path.dirname("/a/b.txt")    # "/a" (폴더 경로만 추출)  
os.makedirs("new/folder")      # 중첩 폴더 한번에 생성  
os.remove("file.txt")           # 파일 삭제  
os.rename("old.txt", "new.txt") # 파일명 변경  
os.environ.get("PATH")         # 환경변수 읽기  

## 5. sys — 인터프리터/시스템 관련
python  
import sys  

sys.argv                # 커맨드라인 인자 리스트 (실행 시 넘긴 값들)

sys.stdin / sys.stdout   # 표준 입출력 스트림  

## 6. time — 시간 측정/지연
python  
import time  

time.time()          # 현재 시각 (1970년 1월 1일부터 초 단위, "epoch time")  
time.sleep(2)         # 2초간 코드 실행 멈춤  
time.strftime("%Y-%m-%d")  # 현재 시간을 포맷팅한 문자열  

start = time.time()   # ... 어떤 작업 ...  
end = time.time()  
print(end - start)   # 걸린 시간(초) 측정에 자주 사용  

💡 팁: datetime과 time은 헷갈리기 쉬운데, **시간 측정(경과 시간)**은 time, 날짜 계산/포맷팅은 datetime을 주로 씁니다

## pandas 라이브러리 사용하기

```python
import pandas as pd
```

```python
# 1. dataframe 생성
# 아래 표와 같은 데이터를 담은 dataframe을 생성하세요.

#표로 보기위한 원시 데이터(딕셔너리)
data = {
    '이름': ['김철수', '이영희', '박민수', '정수진', '최영호', '한미래', '윤지원'],
    '나이': [25, 28, 22, 31, 27, 24, 29],
    '점수': [85, 92, 78, 95, 88, 73, 91]
}
```

```python
#판다스의 데이터프레임객체 생성
df = pd.DataFrame(data)
df.head() # 데이터프레임의 상위 5개 행을 출력
```

```python
average_age = df['나이'].mean() # 나이 컬럼의 평균 계산
print("평균 나이:", int(average_age))
```
