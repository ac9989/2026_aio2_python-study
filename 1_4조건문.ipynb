```python
age = int(input('나이를 입력하세요: '))

if age >= 18:
    print("성인입니다.")
else:
    print("미성년자입니다.")
```

```python
sentence = input("문장을 입력하세요: ")
words = ["파이썬", "코딩", "데이터", "AI"]

found_word = None

for word in words:
    if word in sentence:
        found_word = word
        break

if found_word:
    print(f"입력한 문장에서 '{found_word}' 단어를 찾았습니다!")
else:
    print("입력한 문장에 해당 단어가 없습니다.")
```

```python
score = int(input("점수를 입력하세요: "))
print('pass') if score >= 80 else print('fail')    # 꼭 if문을 줄 바꿔가며 작성할 필요는 없고, 한 줄로 작성 가능하다!
```

```python
num = int(input("숫자를 입력하세요: "))

if num % 2 == 1:
    print("홀수입니다.")
else:
    print("짝수입니다.")
```

```python
# 조건부 표현식
num = int(input("숫자를 입력하세요: "))
print("홀수입니다." if num % 2 == 1 else "짝수입니다.")
```

```python
# else if 문을 사용하여 여러 조건을 처리할 수도 있다.
```

```python
age = int(input("나이를 입력하세요: "))

if age <= 7 or age >= 65:
    print("무료입니다.")
elif age <= 18:
    print("요금은 5,000원입니다.")
else:
    print("요금은 10,000원입니다.")
```

# 중첩 조건문

```python
x, y = 3, -1

if x > 0 : #예외 조건
    if y > 0 :
        print("I 사분면")
    else :
        print("IV 사분면")
else :
    if y > 0 :
        print("II 사분면")
    else :
        print("III 사분면")
```
